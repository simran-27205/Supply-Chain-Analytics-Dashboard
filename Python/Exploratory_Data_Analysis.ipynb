**Feature Engineering**

```python
final_df["delivery_delay"] = (
    final_df["actual_shipping_days"] -
    final_df["scheduled_shipping_days"]
)
```

```python
final_df[["actual_shipping_days",
          "scheduled_shipping_days",
          "delivery_delay"]].head()
```

```python
final_df["delivery_status_new"] = final_df["delivery_delay"].apply(
    lambda x: "Delayed" if x > 0 else "On Time"
)
```

```python
final_df["delivery_status_new"].value_counts()
```

```python
final_df["inventory_value"] = (
    final_df["current stock"] *
    final_df["product_price"]
)
```

```python
final_df["stock_gap"] = (
    final_df["current stock"] -
    final_df["reorder point"]
)
```

```python
final_df["reorder_required"] = final_df.apply(
    lambda row: "Yes"
    if row["current stock"] <= row["reorder point"]
    else "No",
    axis=1
)
```

```python
final_df["reorder_required"].value_counts()
```

```python
def stock_status(row):
    if row["current stock"] <= row["safety_stock"]:
        return "Critical"

    elif row["current stock"] <= row["reorder point"]:
        return "Low Stock"

    else:
        return "Healthy"
```

```python
final_df["stock_status"] = final_df.apply(stock_status, axis=1)
```
